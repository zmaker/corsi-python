## Filtrare i dati

```python
tab
```

```python
#voglio selezionare solo i maschi 
df1 = tab[ tab.sesso == "M" ]
df1
```

```python
tab[ tab.sesso == "F" ]
```

```python
tab[ tab.altezza >= 1.70 ]
```

Posso combinare più espressioni usando degli operatori:
- AND -> &
- OR  -> |
- NOT -> ~

```python
tab[ (tab.altezza >= 1.70) & (tab.altezza < 1.85) ]
```

```python
tab[ (tab.altezza < 1.70) | (tab.altezza > 1.90)]
```

```python
tab[ ~(tab.nome == "Bettina") ]
```

## Elaborare righe e colonne

```python
tab
```

```python
#estraggo una riga
r = tab.loc[2]
r
```

```python
#sovrascrivo/modifico una riga
tab.loc[8] = ["Ugo", "Fracchia", 89.2, 1.55, "1956-07-01", "M", 23.3, "MALE"]
tab
```

```python
#aggiungo una nuova riga
tab.loc[10] = ["Lino", "Banfi", 87.2, 1.65, "1958-07-01", "M", 23.3, "MALE"]
tab
```

```python
#modifico una parte/un campo della riga
tab.at[10, "altezza"] = 1.67
tab
```

```python
#modifico più valori
tab.loc[10, ["peso", "altezza"]] = [85.0, 1.66]
tab
```

```python
#elimino una riga
tab3 = tab.drop(10)
tab3
```

```python
#elimino una colonna
tab4 = tab.drop("genere", axis=1)
tab4
```
