## Pandas - Merge e Join

```python
import pandas as pd
```

```python
#genero un dataframe con un dict
data = {"codice":range(1001, 1006), 'nome':["mela","pera","kiwi","banana","fragola"]}
df1 = pd.DataFrame(data)
df1
```

```python
#genero una seconda tabella con le qta
data = {"codice":range(1001, 1011), 'qta':[10,20,30,40,12,22,44,55,66,77]}
df2 = pd.DataFrame(data)
df2
```

```python
#unisco le due tabelle
df3 = pd.merge(df1, df2, on="codice", how="left")
df3
```

```python
df3 = pd.merge(df1, df2, on="codice", how="right")
df3
```

```python
#aggiungo una riga a df1
df1.loc[5] = {"codice":1100, "nome":"fichi"}
df1
```

```python
df3 = pd.merge(df1, df2, on="codice", how="right")
df3
```

```python
df3 = pd.merge(df1, df2, on="codice", how="inner")
df3
```

```python
df3 = pd.merge(df1, df2, on="codice", how="outer")
df3
```

### Concat
Unire due dataframe con struttura compatibile

```python
data = {
    "codice":[10,20,30],
    "nome":["mela", "pera", "kiwi"],
    "qta":[100, 87, 45]
}
dfa = pd.DataFrame(data)
dfa
```

```python
data = {
    "codice":[100, 200],
    "nome":["patate", "sedano"],
    "qta":[33,45]
}
dfb = pd.DataFrame(data)
dfb
```

```python
dfc = pd.concat([dfa, dfb], ignore_index=True)
dfc
```

```python
dfd = pd.concat([dfa, dfb])
dfd
```

```python
dfd.reset_index(drop=True, inplace=True)
dfd
```
