### File CSV
CSV = Comma Separated Values


```python
import csv
```

```python
with open("dati.csv") as f:
    rows = csv.reader(f, delimiter=",")
    r1 = next(rows)
    print(r1)
    print("")

    for rr in rows:
        #print(rr)
        codice = rr[0]
        prodotto = rr[1]
        quantita = int(rr[2])
        prezzo = float(rr[3])
        print(f"[{codice}] - {prodotto} V:{quantita*prezzo}")
```

Es.: leggi il file csv e calcola il numero totale di oggetti, il valore totale e il valore medio di un oggetto

```python
qta_totale = 0
val_totale = 0

with open("dati.csv") as f:
    rows = csv.reader(f, delimiter=",")
    r1 = next(rows) #scarto le intestazioni nella prima riga

    for rr in rows:
        codice = rr[0]
        prodotto = rr[1]
        quantita = int(rr[2])
        prezzo = float(rr[3])
        print(codice, quantita)

        qta_totale += quantita
        val_totale += (quantita*prezzo)
    
    print(f"Q. Tot: {qta_totale}")
    print(f"V. Tot: {val_totale}")
    print(f"V. Medio: {val_totale/qta_totale:.2f}€")
```
