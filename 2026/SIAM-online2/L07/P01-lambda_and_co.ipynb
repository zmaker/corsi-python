#### lambda expression

```python
# lambda <arg>: <codice>
#
# def nome(<arg>):
#     <codice>

f2 = lambda x: x*x

#def f1(x):
#    return x*x

print(f2(10))
```

```python
for i in range(10):
    test = (lambda x: "P" if (x%2) == 0 else "D")(i)
    print(i, test)
```

```python
m1 = lambda x,y: x*y
print(m1(10,20))
```

### comprehension

```python
lista = [i*2 for i in range(10)]
print(lista)
```
