```python
from openpyxl import load_workbook

#apro il file excel
wb = load_workbook("caseroma.xlsx")

#stampo i fogli di lavoro
print(wb.sheetnames)
```

```python
#lista delle pagine del foglio excel
pagine = wb.sheetnames

#seleziono la prima pagina
print("seleziono la pagina", pagine[0])
ws = wb[pagine[0]]
```

```python
#leggo una cella
contenuto_cella = ws["B2"].value
print(contenuto_cella)
```

```python
#leggo una riga
riga = []
for cl in ws[2]: #tutte le celle della riga 2
    riga.append(cl.value)

print(riga)
```

```python
#modifico una cella
ws["B2"] = "Vaticano"
```

```python
#salvo il file
wb.save("caseroma_mod.xlsx")
```
