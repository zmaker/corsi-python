## statistiche
media, varianza ecc ecc

```python
import statistics as st
voti = [4,5,4,6,7,8,6,7,5,6,9,8,7]
```

```python
#media dei voti
n = st.mean(voti)
print(f"media {n:.2f}")
```

```python
#arrotondo un numero con round
m = round(n, 3)
print(m)
```

```python
#minimo e massimo
n = min(voti)
m = max(voti)
print(f"valori compresi tra {n} e {m}")
```

```python
#mediana
n = st.median(voti)
print("mediana", n)
```

```python
#ordinare i dati 
print(voti)
voti.sort()
print(voti)
```

```python
#mettere in disordine una lista
import random 
random.shuffle(voti)
print(voti)
```
