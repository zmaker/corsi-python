## Filtraggio numeri

```python
n = np.random.randint(0,100, 20)
print(n)

#filtro n ed estraggo i numeri minori di 50
nf = n[ n<50 ]
print(nf)
```

```python
#filtro n e estraggo solo lgi elementi che valgono 2
nf = n[ n==2 ] 
print(nf)
```

```python
#filtro n ed estraggo i numeri diversi da 2 (diverso è !=)
nf = n[ n!=2 ]
print(nf)
```

```python
#stampo solo i numeri pari
npa = n[ (n%2) == 0 ]
print(npa)
```

```python
#numeri compresi in un intervallo 
#combino due espressioni che devono essere vere entrambe -> AND -> &
#numeri tra 20 e 30
nf = n[ (n>=20) & (n<=30) ]
print(nf)
```

```python
#numeri esterni a un intervallo
#numeri maggiori di 40 o numeri minori di 20
# OR -> |
nf = n[ (n<20) | (n>40) ]
print(nf)
```

## Where

```python
n = np.random.randint(0,100, 20)
print(n)
# where applica l'espressione n>40 agli elementi della lista. Se è vera restituisce n altrimenti 0
nn = np.where( n>40, n, 0 )
print(nn)
```

```python
n2 = np.where( (n%2 == 0), 'p', 'd' )
print(n)
print(n2)
print(np.where( (n%2 == 0)))
```

## Arrotondamenti

```python
#genero 10 numeri casuali tra 0 e 1
v = np.random.random(10)
print(v)
```

```python
#li voglio trasformare in numeri tra 0 e 10
v *= 10.0
print(v)
```

```python
#elimino un po' di decimali
q = np.around(v, 2)
print(q)
```

```python
#arrotondo all'intero "superiore"
w = np.ceil(q)
print(w)
```

```python
#arrotondo all'intero inferiore
e = np.floor(q)
print(e)
```

## Esempi

Mi danno una serie di misure (la simuleremo) e di queste fornisco degli indicatori statistici

```python
import numpy as np

#genero i dati delle misure (pesi?)
dati = np.random.random(20) * 100
#riduco i decimali
dati = np.around(dati, 2)
print(dati)

#calcolo gli indici statistici
media = np.average(dati)
dev_std = np.std(dati)
minimo = np.amin(dati)
massimo = np.amax(dati)
mediana = np.median(dati)

print(f"media: {round(media, 2)}\t std: {round(dev_std, 2)}\t min: {minimo}\n max: {massimo}\t mediana: {mediana}")
```
