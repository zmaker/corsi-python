```python
#apro il db e ottengo la connessione
with sqlite3.connect("caffe.db") as conn:
    #creo il cursore
    cur = conn.cursor()
    #passo una query al cursore
    sql = "SELECT * FROM magazzino"
    cur.execute(sql)
    #estraggo i dati con un ciclo for
    for row in cur.fetchall():
        #print(row)
        id, caffe, qta, prz = row #unpacking
        print(f"({id}) - {caffe}\t{qta}\t{prz}")
```

**Ex**: quanti kg di caffè nel magazzino? e quale il valore totale?

```python
qtot = 0
vtot = 0

#apro il db e ottengo la connessione
with sqlite3.connect("caffe.db") as conn:
    #creo il cursore
    cur = conn.cursor()
    #passo una query al cursore
    sql = "SELECT * FROM magazzino"
    cur.execute(sql)
    #estraggo i dati con un ciclo for
    for row in cur.fetchall():
        #print(row)
        id, caffe, qta, prz = row #unpacking
        #print(f"({id}) - {caffe}\t{qta}\t{prz}")

        qtot += qta
        vtot += qta * prz

print("QTA TOT: ", qtot)
print("VAL TOT: ", vtot)
```

**INSERT**: inserire una riga nella tabella

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()

    prod = "real"
    qta = 12
    prz = 3.56

    sql = f"INSERT into magazzino ('prodotto','qta','prezzo') VALUES "\
    f"('{prod}', {qta}, {prz})"
    print(sql)

    cur.execute(sql)
    conn.commit()
```

**UPDATE**: per modificare una riga di database

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    id = 6
    qta = 10
    prz = 7.5
    sql = f"UPDATE magazzino SET qta = {qta}, prezzo = {prz} WHERE mid = {id}"
    cur.execute(sql)
    conn.commit()
```

**DELETE**: per cancellare una riga usando la chiave primaria

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    id = 6
    sql = f"DELETE FROM magazzino WHERE mid = {id}"
    cur.execute(sql)
    conn.commit()
```

## Creare una tabella da codice sql

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    sql = "CREATE TABLE persone ("\
        "pid INTEGER PRIMARY KEY AUTOINCREMENT,"\
        "nome TEXT (20)      NOT NULL"\
        ");"
    cur.execute(sql)
    conn.commit()
```

```python
#elimino la tabella
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    sql = "DROP TABLE persone"
    cur.execute(sql)
    conn.commit()
```
