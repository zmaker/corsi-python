### JSON

importo la libreria json e leggo un fiel per intero

```python
import json

with open("dati.json") as f:
    doc = json.load(f)
    print(doc)
```

```python
with open("dati.json") as f:
    doc = json.load(f)
    id = doc["id"]
    print("id\t: ", id)
    print("type\t: ", doc["type"] )
    print(doc["topping"])
    for el in doc["topping"]:
        print(f"id: {el['id']} - {el['type']}" )
    print(doc["batters"]["batter"])
    for el in doc["batters"]["batter"]:
        print(f"id: {el['id']} - {el['type']}" )
```

Creo un pacchetto dati in json (da python) e lo salvo come file

```python
doc = {"nome":"mario", "cognome":"rossi", "cod":123, "data":"01/01/2001", 
       "materie":[{"materia":"ita", "cod":"01"}, {"materia":"eng", "cod":"02"}, {"materia":"mat", "cod":"03"}]}
with open("materie.json", "w") as f:
    json.dump(doc, f)
```

Interrogazione di un servizio web che pubblica json.

Serve la libreria requests.

pip install requests

```python
# comando per installare la libreria 
# %pip install requests

import requests
```

```python
url = "https://jsonplaceholder.typicode.com/posts/3"

response = requests.get(url)
if response.status_code == 200:
    data = response.json()
    print(data)
    print(data["id"])
    print(data["title"])
else:
    print("errore del server", response.status_code)
```

```python
#ottengo una lista di dati
url = "https://jsonplaceholder.typicode.com/posts"

response = requests.get(url)
if response.status_code == 200:
    doc = response.json()
    for el in doc[:5]: #limita il for ai primi 5 post
        print(el)

else:
    print("errore del server", response.status_code)
```

```python
#uso try except per intercettare eventuali errori

url = "https://jsonplaceholder.typicode.com/posts/3"

try:
    response = requests.get(url, timeout=5)
    response.raise_for_status() #se ci sono errori, genera eccezioni

    if response.status_code == 200:
        data = response.json()
        print(data)
        print(data["id"])
        print(data["title"])
    else:
        print("errore del server", response.status_code)
except:
    print("errore ricezione dati")
```
