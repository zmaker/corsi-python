# Dictionary
Strutture dati in cui memorizzo delle *coppie* **chiave-valore**


```python
listino = {"mele":12, "pere":23, "kiwi":45}
print(listino)
```

Estraggo un valore dal dict usando la **chiave**

```python
el = listino["pere"]
print(el)

el2 = listino.get("kiwi")
print(el2)
```

Posso modificare un valore sempre utilizzando la sua chiave

```python
listino["mele"] = 99
print(listino)
```

Come aggiungere nuovi elementi

```python
listino["fragole"] = 123
print(listino)
```

Posso eliminare un elemento con pop(*chiave*)

```python
listino.pop("fragole")
listino
```

```python
if "patate" in listino:
    listino.pop("patate")
else:
    print("non presente")
```

Oltre a pop() posso usare **del** che ha una sintassi molto particolare

```python
del listino["kiwi"]
print(listino)
```

Posso svuotare un dizionario con clear()

```python
listino.clear()
print(listino)
```

Accesso agli elementi 

```python
listino = {"mele":12, "pere":23, "kiwi":45}
```

```python
print(listino.keys())
```

```python
for k in listino.keys():
    print(k, listino[k])
```

```python
for v in listino.values():
    print(v)
```

```python
for i in listino.items():
    k, v = i #unpacking perché i è una tupla
    print("k:", k, " val:", v)
```

Copia di dizionari

```python
listino2 = listino #non è una vera copia!!!
print("l1", listino)
print("l2", listino2)
```

```python
listino["mele"] = 0
print(listino)
```

```python
print(listino2)
```

La copia "deep-copy" o "full-copy" si ottiene con .copy()

```python
listino3 = listino.copy()
print(listino)
print(listino3)
```

```python
listino["mele"] = 200
```

```python
print(listino)
print(listino3)
```
