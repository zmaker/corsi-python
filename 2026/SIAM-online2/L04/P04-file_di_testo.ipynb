```python
with open("file1.txt") as f:
    pass
```

```python
with open("file1.txt") as f:
    print(f.read())
```

```python
with open("file1.txt") as f:
    print(f.read(5))
    f.seek(0)
    print(f.read(5))
```

```python
with open("file1.txt") as f:
    line = f.readline()
    print(line, end='')
    line = f.readline()
    print(line, end='')
```

```python
with open("file1.txt") as f:
    ll = f.readlines()
    print(ll)
```

```python
#stampo tutte le righe usando un ciclo for
with open("file1.txt") as f:
    for riga in f:
        print(riga, end='')
```

**scrivere file**
Per aprire un file in scrittura usao il flag "w" (write). In questo modo il file è distrutto e ricreato ogni volta.

```python
with open("file3.txt", "w") as f:
    f.write("pane\n")
    f.write("latte\n")
    f.write("burro\n")
```

Se voglio mantenere il file e solo aggiungere dati, uso "append" "a"

```python
with open("file4.txt", "a") as f:
    f.write("ciao!\n")
```

Esempio di datalogger

```python
import datetime
import random

with open("datalog.txt", "a") as f:
    t1 = datetime.datetime.now()
    f.write(str(t1))
    sens = random.randint(20, 50)
    f.write(";"+str(sens)+"\n")
```

## Libreria os

```python
import os
if os.path.exists("datalog.txt"):
    print("Il file esiste: lo cancello!")
    os.remove("datalog.txt")
else:
    print("il file non esiste: lo creo")
    f = open("datalog.txt", "x")
    f.close()
```
