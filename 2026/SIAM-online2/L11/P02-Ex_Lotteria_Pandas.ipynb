### Lotteria Pandas

Simulo delle estrazioni del Lotto. 
Voglio trovare i numeri ritardatari.

```python
import numpy as np

numeri = np.arange(1, 91)
numeri
```

```python
#mescolo le palline
np.random.shuffle(numeri)
numeri
```

```python
#creo una lista in cui mettere le estrazioni
estrazioni = np.array([])
```

```python
for i in range(20):
    #rimescolo
    np.random.shuffle(numeri)
    #prendo i primi 5
    estratti = numeri[:5]
    print(estratti)

    #aggiungo a estrazioni
    estrazioni = np.append(estrazioni, [estratti])
    
estrazioni
```

```python
#riarrangio i dati
estrazioni = estrazioni.reshape(20, 5)
estrazioni
```

```python
import pandas as pd

dates = pd.date_range("20260101", periods=20)
dates
```

```python
df = pd.DataFrame(data=estrazioni, index=dates)
df
```

```python
ritardatari = {}
for i in range(len(df)):
    riga = df.iloc[i]
    for num in riga:
        gg = len(df)-i
        ritardatari[num] = gg
ritardatari
    
```

```python
df = pd.DataFrame(list(ritardatari.items()), columns=['numero', "ritardo"])
df.head(4)
```

```python
df = df.sort_values(by="ritardo", ascending=False)
df.head(5)
```
