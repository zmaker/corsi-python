Creo una classe in cui aggiungo delle proprietà. Le proprietà è bene aggiungerle nel costruttore

```python
class Banana():    
    def __init__(self):
        self.peso = 1000
        self.colore = "giallo"
        self.lunghezza = 20        
```

```python
b1 = Banana()
b1.lunghezza = 21
b1.peso = 256
print("banana:", b1.colore, b1.peso, b1.lunghezza)
```

```python
class Kiwi():    
    def __init__(self):
        self.peso = 100
        self.colore = "brown"
        self.lunghezza = 7    
    
    def getPeso(self):
        return self.peso
    
    def setPeso(self, p):
        if (p < 0):
            self.peso = 0
        else:
            self.peso = p
```

```python
k1 = Kiwi()
k1.setPeso(25)
print("kiwi:", k1.getPeso())
print(k1)
```

#### str()
Rappresentazione dell'oggetto come stringa

```python
class Pesca():
    def __init__(self):
        self.peso = 0
        self.colore = "giallo"
    
    def __str__(self):
        return f"Pesca ({self.colore})"
```

```python
p1 = Pesca()
print(p1)
```

### Parametri del costruttore

```python
class Patata():
    def __init__(self, weight=100, color="marrone"):
        self.peso = weight
        self.colore = color
    def __str__(self):
        return f"Patata - {self.colore} {self.peso}"

p1 = Patata(120, "marroncina")
print(p1)
p2 = Patata()
print(p2)
p3 = Patata(75, "gialla")
print(p3)
```

#### distruttore

```python
class Punto():
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x}, {self.y})"
    
    def __del__(self):
        print("bye bye")

a = Punto(10, 20)
o = Punto()
print(a)
print(o)

# forzo la distruzione dell'oggetto in modo esplicito
del a
```

Fatto strano sulle proprietà di un oggetto e sul perché sia meglio raccoglierle tutte nel costruttore

```python
class Test():
    def __init__(self):
        self.a = 0

    def f1(self):
        self.b = 100
```

```python
t1 = Test()
t1.a = 10
t1.f1()
t1.b = 20
print(t1.a, t1.b)
```

Esempio di classe con proprietà def non nel costruttore

```python
class Zucchina():
    def __init__(self, peso, lun, col="verde"):
        self.peso = peso
        self.lun = lun
        self.col = col
    
    def __str__(self):
        txt = f"Z: {self.peso}gr, {self.lun}cm, {self.col}"
        return txt
    
    def sbuccia(self):
        self.peso = self.peso * 0.9
    
    def affetta(self, ft):
        #proprietà definita non nel costruttore! (pericoloso)
        self.fette = ft

z1 = Zucchina(10, 23)
print(z1)
z1.sbuccia()
print(z1)
z1.affetta(5)
print(z1.fette)
```

## Esercizio Biblioteca

```python
class Book():
    def __init__(self, titolo="-", autore="?"):
        self.titolo = titolo
        self.autore = autore
    def __str__(self):
        return f"L: {self.titolo} di {self.autore}"
```

```python
class Biblioteca():
    def __init__(self):
        self.scaffale = []
    
    def addBook(self, titolo, autore):
        l = Book(titolo, autore)
        self.scaffale.append(l)
    
    def __str__(self):
        txt = ""
        for i, b in enumerate(self.scaffale):
            txt += str(i+1) + " - " + str(b) + "\n"
        return txt
    
    def search(self, titolo):
        ans = Book()
        for b in self.scaffale:
            if (b.titolo == titolo):
                #trovato il libro
                ans = b
                break
        return ans

```

```python
b = Biblioteca()
# aggiungo libri
b.addBook("Promessi Sposi", "Manzoni")
b.addBook("Divina Commedia", "Dante")
b.addBook("Python per tutti", "Monty Python")


#stampo l'elenco dei libri presenti
print(b)

#cerco un titolo
book = b.search("Divina Commedia")
print(book)

book = b.search("Milano Segreta")
print(book)
```
