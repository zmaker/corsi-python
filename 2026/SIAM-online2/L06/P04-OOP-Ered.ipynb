## Ereditarietà in Python

```python
class Auto():
    def __init__(self, colore, anno):
        self.colore = colore
        self.anno = anno
        self.vel = 0

    def setVelocita(self, spd):
        self.vel = spd
    
    def beep(self, suono):
        return f"{suono})))"
    
    def __str__(self):
        return f"Auto - c:{self.colore} - {self.anno} - v:{self.vel}Km/h"
```

```python
a0 = Auto("blu", 2012)
print(a0.beep("honk"))
a0.setVelocita(45)
print(a0)
```

```python
class Fiat500(Auto):
    pass
```

```python
a1 = Fiat500("gialla", 1976)
a1.setVelocita(100)
print(a1)
```

```python
class Ferrari(Auto):
    def __init__(self, colore, anno, turbo):
        super().__init__(colore, anno)
        self.turbo = turbo
    
    def beep(self, suono):
        return f"{suono}!!!!!!"
    
    def __str__(self):
        return f"Ferrari - c:{self.colore} - {self.anno} - v:{self.vel}Km/h turbo:{self.turbo}"
```

```python
f1 = Ferrari("rosso", 2000, True)
print(f1)
print(f1.beep("BOOOOOO"))
```
