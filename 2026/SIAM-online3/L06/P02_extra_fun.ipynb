# Approfondimenti sulle funzioni

Parametri nominali

```python
def stampaPunto(x, y, colore):
    print(f" . {x};{y} c:{colore} ")
```

```python
stampaPunto(10, 10, "blu")
```

```python
stampaPunto(y=23, colore='rosso', x=10)
```

Parametri con valori di default

```python
def stampaLinea(xi=0, yi=0, xf=100, yf=100, colore='nero'):
    print(f"linea {colore} da ({xi};{yi}) a ({xf};{yf})")

stampaLinea(1,1, 5,6, "rossa")
stampaLinea()
stampaLinea(colore="giallo", xf=20, xi=12, yf=4, yi=5)
```

Funzioni che restituiscono valori multipli

```python
def calc(a, b):
    s = a+b
    p = a*b
    return s, p #packing

x = calc(10, 20)
print(x)

somma, prodotto = calc(23,34) #unpacking
print(somma, prodotto)
```

Parametri arbitrari

```python
print("ciao")
print("a", "b", 2, "c")
```

```python
def cugini(*nome):
    #print(len(nome), nome)
    for cg in nome:
        print(cg, end=" ")
    print()

cugini()
cugini("luigi")
cugini("luigi", "mario", "anna")
```

Creare una funzione che somma un numero arbitrario di numeri

```python
def sumx(*n):
    if len(n) == 0:
        return 0
    elif len(n) == 1:
        return n[0]
    else:
        somma = 0
        for el in n:
            somma += el
        return somma

print(sumx(2,3))
print(sumx(1,2,3))
print(sumx(10, 30, 100))
#casi particolari
print(sumx())
print(sumx(1))
```
