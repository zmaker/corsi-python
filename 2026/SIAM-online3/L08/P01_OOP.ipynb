Le proprietà della classe

```python
class Banana():
    def __init__(self):
        self.colore = "giallo"
        self.peso = 100
        self.lunghezza = 13
```

```python
b1 = Banana()
print(b1.colore)
b1.peso = 120
print(b1.peso)
```

```python
class Kiwi():
    def __init__(self):
        self.peso = 70
        self.colore = "marrone"
        self.lunghezza = 0
    
    def setPeso(self, p):
        if p > 0:
            self.peso = p
        else:
            self.peso = 0
```

```python
k1 = Kiwi()
k1.setPeso(100)
print(k1.peso)
print(k1)
```

Parametri del costruttore

```python
class Patata():
    def __init__(self, c, p=0):
        self.peso = p
        self.colore = c 
    def __str__(self):
        return f"Patata {self.colore} - {self.peso}"

p1 = Patata("gialla", 120)
print(p1)
p1.colore = "rossa"
print(p1)

p2 = Patata("marroncina")
print(p2)
#p3 = Patata()
p4 = Patata(p=12, c="bianca")
print(p4)
```

**Esercizio Biblioteca OOP**

Useremo due entità: libro e biblioteca

```python
class Book():
    def __init__(self, titolo="-", autore="?"):
        self.autore = autore
        self.titolo = titolo

    def __str__(self):
        return f"L: {self.titolo} - {self.autore}"
```
