# Funzioni
funzioni in Python

la funzione saluta non ha parametri e non restituisce nulla. SI chiama anche **procedura**

```python
def saluta():
    # corpo della funzione
    # questa funzione vi saluta
    print("ciao!")

saluta()
```

```python
saluta()
```

Ora una funzione che riceve un parametro:

(passaggio per valore)

```python
def salutaConNome(nome):
    print("ciao,", nome)

salutaConNome("mario")
a = "Luigi"
salutaConNome(a)
```

Funzione che restituisce un risultato

```python
def somma(a, b):
    s = a + b
    return s
```

```python
print(somma(10, 20))

z = somma(2,3)
print(z)

n = 10
m = 23
print(somma(n,m))
```
