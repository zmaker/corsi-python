# File di testo in python
come aprire, leggere e scrivere file di testo in python

Apro un file e lo chiudo

```python
f = open("file1.txt")
#operazioni sul file
f.close()
```

Di seguito la versione compatta del codice per aprire e chiudere i file. Useremo **with** che in automatico esegue il *close()*.

```python
with open("file1.txt") as f:
    pass
```

Proviamo a leggere il file

```python
with open("file1.txt") as f:
    print(f.read())
```

```python
with open("file1.txt") as f:
    # leggo 5 caratteri
    print(f.read(5))
    # porto la testina all'inizio del file
    f.seek(0)
    # leggo 5 caratteri
    print(f.read(5))
```

Leggo il file riga per riga

```python
with open("file1.txt") as f:
    linea = f.readline()
    print(1, linea)
    linea = f.readline()
    print(2, linea)
```

```python
with open("file1.txt") as f:
    ll = f.readlines()
    #print(ll)
    for linea in ll:
        print(linea, end='')
    
```

```python
with open("file1.txt") as f:
    for riga in f:
        print(riga, end='')
```

esempio di apertura file con percorso assoluto

```python
with open("C:\\temp\\file2.txt") as f:
    for riga in f:
        print(riga, end='')
```

## Scrittura file

W apre un file in scrittura e se non lo trova, lo crea 

```python
with open("file3.txt", "w") as f:
    f.write("pane\n")
    f.write("latte\n")
    f.write("burro\n")
```

file aperto in append

```python
with open("file4.txt", "a") as f:
    f.write("ciao\n")
```

esempio datalogger

```python
import random
import datetime
with open("datalog.txt", "a") as f:
    t1 = datetime.datetime.now()
    f.write(str(t1))
    sens = random.randint(20, 40)
    f.write(";"+str(sens)+"\n")
```

# Libreria OS

```python
import os
if os.path.exists("datalog.txt"):
    print("trovato")
    os.remove("datalog.txt")
else:
    print("file non esiste")
    f = open("datalog.txt", "x")
    f.close()
```
