# lambda function

```python
def f1(x):
    return x*x

print(f1(3))
```

```python
# lambda expression:
# lambda <arg>: <codice>

f2 = lambda x: x*x
print(f2(10))
```

```python
for i in range(5):
    test = ( lambda x: ( "P" if (x%2) == 0 else "D") )(i)
    print(i, test)
```

```python
m1 = lambda n,m: n*m
print(m1(10,20))
```

# generatori
un generatore è una funzione che mantiene memoria del suo stato interno tra una chiamata e l'altra

```python
def quad(valore = 0):
    while(True):
        yield f"il quadrato di {valore} è {valore*valore}"
        valore += 1

g = quad()

next(g)
```

```python
def somma(a, b):
    s = a + b
    return s

print(somma(2,3))
```

# Comprehension

```python
lista = [ i*2 for i in range(10) ]
print(lista)
```
