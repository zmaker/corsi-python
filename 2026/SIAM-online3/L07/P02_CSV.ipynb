# CSV in python
CSV = Comma Separated Values

```python
import csv
```

```python
with open("dati.csv") as f:
    rd = csv.reader(f, delimiter=",")
    r1 = next(rd)
    print(r1)
    print("---")

    qta_totale = 0

    for riga in rd:
        #print(riga)
        codice = riga[0]
        prodotto = riga[1]
        qta = int(riga[2])
        prezzo = float(riga[3])

        qta_totale += qta #qta_totale = qta_totale + qta
        print(f"cod: {codice} - {prodotto} Val: {qta*prezzo}")
    
    print("qta totale: ", qta_totale)
```

Scrittura file CSV tramite libreria

```python
with open("magazzino.csv", "w") as f:
    wr = csv.writer(f, delimiter=",", lineterminator="\n")
    wr.writerow( ['cod', 'caffe', 'qta', 'prz'] )
    wr.writerow( ['10', 'java', 100, 0.99] )
    wr.writerow( ['20', 'moka', 120, 1.99] )
    wr.writerow( ['30', 'arabica', 230, 2.99] )
```
