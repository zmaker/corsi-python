## indici e posizioni

due concetti:
- chiave della riga -> **loc()**
- posizione della riga --> **iloc()**

```python
df3 = df.sort_values(by='cognome', ascending=True)
df3
```

```python
# estraggo per "chiave"
r1 = df3.loc[0]
r1
```

```python
# estraggo per posizione
r2 = df3.iloc[0]
r2
```

```python
r2["nome"]
```

# filtrare dati

```python
df
```

```python
df4 = df[ df.sesso == "M" ]
df4
```

```python
df5 = df4[ df4.altezza >= 1.70 ]
df5
```

Posso combinare più espressioni con:
- AND -> &
- OR  -> |
- NOT -> ~

```python
df6 = df[ (df.altezza >= 1.70) & (df.altezza <= 1.80) ]
df6
```

```python
#estraggo tutti i nomi diversi da Puccio
df7 = df[ ~( df.nome == "Puccio" ) ]
df7
```

```python
#modifica o sovrascrittura di una riga
df.loc[8] = ["Fracchia","Ugo", "1956-07-01", 89.2, 1.55, "M", 0, "male"]
df
```

```python

df = df.copy() #è una specie di commit
df.loc[14] = ["Fantozzi","Ugo", "1956-07-01", 89.2, 1.55, "M", 0, "male"]
df
```
