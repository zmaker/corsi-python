## Filtraggi

```python
n1 = np.random.randint(0, 100, 20) #20 numeri a caso tra 0 e 100
print(n1)

nf = n1[ n1<50 ]
print(nf)
```

```python
#numeri uguali a 44
n2 = n1[ n1 == 44 ]
print(n2)
```

```python
#solo i numeri pari
n3 = n1[ (n1%2) == 0 ]
print(n3) 
```

```python
#numeri in un intervallo
n4 = n1[ (n1 > 20) & (n1 < 40)  ] # & = AND, | = OR
print(n4)
```

## where

```python
n5 = np.where( n1>50, n1, 0 )
print(n5)
```

```python
n6 = np.where( (n1%2 == 0), 'p', 'd' )
print(n6)
```
