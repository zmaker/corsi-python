# JSON

```python
import json

with open("ciambelle.json") as f:
    doc = json.load(f)
    print(doc)
```

```python
#accesso alle parti interne del json
with open("ciambelle.json") as f:
    doc = json.load(f)
    codice = doc["id"]
    print(codice)
    print("ppu:", doc["ppu"])
    print("nome:", doc["name"])

    for el in doc["topping"]:
        print(el)
    
    for im in doc["batters"]["batter"]:
        print(im["id"], im["type"])
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts/4"

resp = requests.get(url)
if resp.status_code == 200:
    dati = resp.json()
    print("post n.:", dati["id"])
    print(dati["title"])
    print("------------")
    print(dati["body"])
else:
    print("errore del server")
```

```python
url = "https://jsonplaceholder.typicode.com/posts/4"

try:
    resp = requests.get(url, timeout=5)
    resp.raise_for_status()

    if resp.status_code == 200:
        dati = resp.json()
        print("post n.:", dati["id"])
        print(dati["title"])
        print("------------")
        print(dati["body"])
    else:
        print("errore del server", resp.status_code)
except:
    print("impossibile leggere i dati")
```
