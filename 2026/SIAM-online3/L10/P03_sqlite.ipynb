Estrazione dati con SELECT

```python
with sqlite3.connect("caffe.db") as conn:
    #creo un cursore 
    cur = conn.cursor()
    #passo una query 
    #sql = "SELECT * FROM magazzino WHERE qta >= 80 order by prodotto"
    sql = "SELECT * FROM magazzino ORDER BY mid"
    cur.execute(sql)
    print("id  - nome\tqta\tprz ")
    print("---------------------------------")
    for riga in cur.fetchall():
        #print(riga)
        cid,name,q,p = riga
        print(f"({cid}) - {name}\t{q}\t{p}") 
```

Inserimento riga nel database

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    prod = "green-bean"
    qta = 98
    prz = 3.45
    
    sql = f"INSERT INTO magazzino ('prodotto','qta','prezzo') VALUES "\
        f"('{prod}',{qta},{prz})"
    print(sql)
    cur.execute(sql)

    conn.commit()
```

Modifica di una riga 

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    q = 11
    p = 22
    mid = 8
    sql = f"UPDATE magazzino SET qta={q}, prezzo={p} WHERE mid = {mid}"
    print(sql)
    cur.execute(sql)
    conn.commit()
```

Cancellare una riga

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    id = 8
    sql = f"DELETE FROM magazzino WHERE mid = {id}"
    cur.execute(sql)
    print(sql)
    conn.commit()
    #conn.rollback()
```

Creo una tabella con il DDL (Data Definition Language)


```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    sql = "CREATE TABLE test2 ("\
        "id_persona INTEGER   PRIMARY KEY AUTOINCREMENT, "\
        "nome       TEXT (20) NOT NULL, "\
        "cognome    TEXT (50) NOT NULL)"
    cur.execute(sql)
    conn.commit()
```

Elimino una tabella

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    sql = "DROP TABLE test2"
    cur.execute(sql)
    conn.commit()
```

Ottenere i nomi delle tabelle

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    sql = "SELECT name FROM sqlite_master WHERE type='table'"
    cur.execute(sql)
    for el in cur.fetchall():
        print(el)
```
