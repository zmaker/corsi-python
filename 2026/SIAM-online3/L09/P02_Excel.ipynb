```python
from openpyxl import load_workbook

#apro il file excel
wb = load_workbook("caseroma.xlsx")

#stampo i fogli di lavoro
print(wb.sheetnames)
```

```python
#salvo i nomi delle pagine in una lista
pages = wb.sheetnames

print("seleziono la pagina: ", pages[0])
ws = wb[pages[0]]
```

```python
contenuto = ws["B2"].value
print(contenuto)
```

```python
#estraggo una riga
riga = []
for cl in ws[2]:
    riga.append(cl.value)
print(riga)
```

```python
#scrivo in una cella
ws["B2"] = "Vaticano"
```
