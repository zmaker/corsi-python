**Esercizio Biblioteca OOP**

Useremo due entità: libro e biblioteca

```python
class Book():
    def __init__(self, titolo="-", autore="?"):
        self.autore = autore
        self.titolo = titolo

    def __str__(self):
        return f"L: {self.titolo} - {self.autore}"
```

```python
class Biblioteca():
    def __init__(self):
        self.scaffale = []     

    def addBook(self, titolo, autore):
        libro = Book(titolo, autore)
        self.scaffale.append(libro)
    
    def __str__(self):
        txt = ""
        for bk in self.scaffale:
            txt = txt + str(bk) + "\n"
        return txt
    
    def search(self, titolo_da_cercare):
        ans = Book()
        for bk in self.scaffale:
            if bk.titolo == titolo_da_cercare:
                #libro trovato
                ans = bk
                break
        return ans
```

```python
b = Biblioteca()
#aggiungere dei libri
b.addBook("Promessi Sposi", "Manzoni")
b.addBook("Divina Commedia", "Dante")
b.addBook("Guida galattica autostoppisti", "Douglas Adam")

#stampo elenco libri
print(b)

#cerco un titolo
risultato = b.search("Divina Commedia")
print(risultato)

risultato = b.search("Milano Segreta")
print(risultato)
```

# Ereditarietà

```python
class Auto():
    def __init__(self, colore, anno):
        self.color = colore
        self.year = anno
        self.speed = 0
    
    def setVelocita(self, velocita):
        self.speed = velocita

    def beep(self, suono):
        return f"{suono}.)))"
    
    def __str__(self):
        return f"Auto - c:{self.color} - {self.year} - v:{self.speed}Km/h"
```

```python
a0 = Auto("blu", 2012)
print(a0.beep("honk"))
a0.setVelocita(45)
print(a0)
```

```python
class Fiat500(Auto):
    pass
```

```python
a1 = Fiat500("rossa", 1976)
a1.setVelocita(100)
print(a1)
```

```python
class Ferrari(Auto):
    def __init__(self, colore, anno=2000, turbo=True):
        super().__init__(colore, anno)
        self.turbo = turbo
        
    def beep(self, suono):
        return f"{suono}!!!!!!!"
    
    def saluta(self):
        print("ciao")

    def __str__(self):
        return f"Ferrari {self.speed} {self.turbo}"
```

```python
a2 = Ferrari("rosso", 2000, True)
print(a2.beep("BOOOOOOP"))
print(a2)
a2.saluta()

a3 = Ferrari("rosso")
print(a3)
```
