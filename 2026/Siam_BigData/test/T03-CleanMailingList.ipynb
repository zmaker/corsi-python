## Pulizia mailing list

```python
import pandas as pd

df = pd.read_csv("../dataset/dati_mailing_list.csv", delimiter=",")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.dtypes
```

```python
#rename colonne
df.rename(columns={"__Email_Entered": "email"}, inplace=True)
df.rename(columns={"First name": "nome"}, inplace=True)
df.rename(columns={"_CREDIT": "credito"}, inplace=True)
df.rename(columns={"Surname": "cognome"}, inplace=True)
df.rename(columns={"Age": "eta"}, inplace=True)
df.rename(columns={"Gender": "genere"}, inplace=True)
df.head()
```

```python
df2 = df[["nome", "cognome", "email", "eta", "genere", "credito"]].copy()
df2
```

```python

df2["nome"] = df2["nome"].str.strip().str.title()
df2["cognome"] = df2["cognome"].str.strip().str.title()
df2
```

elimino le righe con email no valida

```python
import re

email_regex = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
```


.str.match(...) → verifica la regex

na=False → considera non valide le email NaN

il filtro elimina automaticamente le righe non valide

```python
df3 = df2[df2["email"].str.match(email_regex, na=False)].copy()
df3
```

elimno duplicati


```python
#più colonne
#df = df.drop_duplicates(subset=["Nome", "email"])
#quale tengo...
#df = df.drop_duplicates(subset=["email"], keep="first")  # default
#df = df.drop_duplicates(subset=["email"], keep="last")

df4 = df3.drop_duplicates().copy()
df4
```

inserisco un default per l'eta

```python
media_eta = df4["eta"].mean()

df4["eta"] = df4["eta"].fillna(media_eta).copy()
df4
```

elimino righe con credito < di 0

```python
df5 = df4[df4["credito"] > 0].copy()
df5
```

specifico che l'indice è email:


```python
if df["email"].is_unique:
    df6 = df5.set_index("email").copy()
```

```python
df6
```

```python
df6.to_csv("ml-output.csv", index=True)
```
