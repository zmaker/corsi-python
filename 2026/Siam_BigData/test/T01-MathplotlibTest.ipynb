### matplotlib test


```python
import matplotlib.pyplot as plt
```

Grafico semplice fornendo una serie di dati

```python
y = [2,3,4,5,5,6,4,5,6]
plt.plot(y)
# definisce i limiti degli assi x e y
plt.axis((0, len(y)-1, min(y)-1, max(y)+1))
plt.show()
```

Con due serie numeriche
* visualizzo punti rossi con ro (red e o per punto)
* The default format string is 'b-', which is a solid blue line

```python
plt.plot([1, 2, 3, 4], [1, 4, 9, 16], 'ro')
plt.axis((0, 5, 0, 20))
plt.show()
```

Visualizzare più serie numeriche

```python
import numpy as np

# da 0 a 5 con passo di 0.2
t = np.arange(0., 5., 0.2)

# red dashes, blue squares and green triangles
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3, 'g^')
plt.show()
```

```python
plt.figure()
plt.subplot(211) #grafico parziale su 2 colonne e 1 riga - in prima posizione 1
plt.plot([1, 2, 3, 4], [1, 4, 9, 16], 'k')

plt.subplot(212) #grafico parziale su 2 colonne e 1 riga - in seconda posizione 2
plt.plot([1, 2, 3, 4], [3,4,6,5], 'r--')
plt.show()
```
