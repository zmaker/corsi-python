### stat e grafici


```python
voti = [4,5,6,5,6,7,6,7,8,7,7,8,9,8,7,5,3,6,7,8,9]
```

```python
import statistics as st
```

```python
st.mean(voti)
```

```python
st.median(voti)
```

```python
max(voti)
```

```python
min(voti)
```

```python
st.variance(voti)
```

```python
st.stdev(voti)
```
