### Database (sqlite3)

```python
import sqlite3
```

Creo un db vuoto

```python
conn = sqlite3.connect("archivio.db")
conn.close()
```

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    cur.execute("select * from magazzino")
    for row in cur.fetchall():
        #print(row)
        mid, prod, qta, prz = row #unpacking dei dati
        print(mid, prod , qta, prz, (qta*prz), sep='\t')
```

Inserisco una riga nel database

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()

    mid = 4
    prodotto = "supermoka" 
    prezzo = 56.2
    qta = 89
    sql = f"insert into magazzino "\
    f"(mid, caffe, qta, prezzo) values "\
    f"({mid}, '{prodotto}', {qta}, {prezzo})"
    print(sql)
    cur.execute(sql)
    conn.commit()
```

modifico una riga

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    mid = 4
    sql = f"update magazzino set qta = 200 where mid = {mid}"
    cur.execute(sql)
    conn.commit()
```

Creo tabella

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    
    sql = f"CREATE TABLE magazzino2 ("\
    "mid    INTEGER PRIMARY KEY NOT NULL,"\
    "caffe  VARCHAR (20)   NOT NULL,"\
    "qta    INTEGER default 0,"\
    "prezzo DECIMAL (5, 2) )"
    cur.execute(sql)
    conn.commit()
```

Cancello tabella

```python
with sqlite3.connect("caffe.db") as conn:
    cur = conn.cursor()
    sql = f"DROP TABLE magazzino2"
    cur.execute(sql)
    conn.commit()
```
