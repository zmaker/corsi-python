## Pandas - Join/merge

Le join di tabelle si fanno con merge che prende come tipo di join: outer, inner , left e right

```python
import pandas as pd
```

```python
data = {
    'codice':range(1001, 1006),
    'nome':['mela', 'pera', 'banana', 'kiwi', 'fragola']
}
df_prodotti = pd.DataFrame(data)
df_prodotti
```

```python
data = {
    'codice':range(1001, 1011),
    'qta':[10,20,30,40,50,60,70,80,90,99]
}
df_magazzino = pd.DataFrame(data)
df_magazzino
```

```python
df1 = pd.merge(df_prodotti, df_magazzino, on="codice", how="left")
df1
```

```python
df2 = pd.merge(df_prodotti, df_magazzino, on="codice", how="right")
df2
```

## concat
Unire due dataframe con la stessa forma (numero e tipo di colonne) 

```python
data = {
    'codice':[10,20,30],
    'nome':['kiwi', 'banana', 'fragola'],
    'qta':[100,50,77]
}
dfa = pd.DataFrame(data)
dfa
```

```python
data = {
    'codice':[40,50],
    'nome':['patate', 'carote'],
    'qta':[55,66]
}
dfb = pd.DataFrame(data)
dfb
```

```python
dfc = pd.concat([dfa, dfb], ignore_index= True)
dfc
```

## Raggruppamenti - GroupBy

```python
data = {
    'codice':[10,20,30,40, 50, 60],
    'nome':['kiwi', 'banana', 'fragola', 'mela', 'pera', 'ananas'],
    'qta':[100,50,77,23,34,45],
    'mag':[1,1,0,0,1,0]
}
dfp = pd.DataFrame(data)
dfp
```

```python
dfp.groupby('mag').groups
```

```python
dfp.groupby("mag").get_group(0)
```

```python
dfp.groupby("mag")['qta'].sum()
```

```python
dfp.groupby("mag")['qta'].describe()
```
