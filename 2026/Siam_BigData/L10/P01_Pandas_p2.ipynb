## Filtrare dati di un dataframe

```python
df = pd.read_csv("persone.csv", delimiter=",")
df
```

```python
df1 = df[df.sesso=='M']
df1
```

```python
df[df.sesso == 'F']
```

```python
df[df.altezza > 1.80]
```

posso combinare più espressioni di selezione usando:

* & per l'AND
* | per l'OR
* ~ per il NOT

```python
df[ (df.altezza > 1.80) & (df.sesso == 'F') ]
```

```python
df[ (df.altezza < 1.70) | (df.altezza > 1.85)]
```

```python
df[ ~(df.nome == 'Tina')]
```

## manipolazione di righe e colonne 

```python
#estraggo una riga usando la sua chiave o indice
row = df.loc[5]
#type(row)
row
```

```python
#estraggo una riga usando la sua posizione
row = df.iloc[0]
row
```

```python
df = df.sort_values(["cognome"])
df
```

```python
row = df.iloc[0]
row
```

```python
#modifico una riga usando l'indice/chiave
df.loc[8] = ["Pier", "Udinesi", 60.0, 1.90, "1993-05-15", 'M']
df
```

```python
df.loc[10] = ["Mario", "Rossi", 75.5, 1.76, "1999-01-01", 'M']
df
```

```python
# modifco solo un campo di una riga
df.at[ 10, "peso"] = 90
df
```

```python
# modifco più campi di una riga usando l'indice 
df.loc[ 10, ["peso", "altezza"]] = [91, 1.77]
df
```

```python
#Elimino una riga
df.drop(10)
```
