# Operatore ternario

```python
n = 10
print("PARI" if (n%2) == 0 else "DISPARI")
```

```python
n = 3
print("PARI" if (n%2) == 0 else "DISPARI")
```

```python
n = 5
if (n%2) == 0:
    print("PARI")
else:
    print("DISPARI")
```

Lambda Function

```python
perdue = lambda x : x*2
m = perdue(4)
print(m)
```

```python
def op(a, b, operazione):
    return operazione(a, b)

```

```python
res = op(10,20, lambda x,y : x+y)
print(res)
res = op(10,20, lambda p,q : p*q)
print(res)
```

Generatori


```python
def serieq(x):
    while(True):
        yield x*2
        x += 1

g = serieq(2)
print(next(g))
print(next(g))
print(next(g))
```

# comprehension

```python
n = [ i*2 for i in range(10) ]
n
```

```python
n = [ i*i for i in range(-5,5) ]
n
```

```python
n = { i*i for i in range(-5,5) }
n
```

```python
n = tuple( i*2 for i in range(10) )
n
```

```python
n = { f"cod{i}":i*i for i in range(5) }
n
```
