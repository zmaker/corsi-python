# OOP Ereditarietà

```python
class Auto:
    def __init__(self, colore, anno):
        self.colore = colore
        self.anno = anno
        self.vel = 0
    
    def setSpeed(self, vel):
        self.vel = vel
    
    def beep(self, sound):
        return f"Questa auto fa...: {sound}"
    
```

creo delle classi specializzate

```python
class Fiac500(Auto):
    pass
```

```python
f = Fiac500("rosso", 1956)
print(f.colore)
f.setSpeed(100)
print(f.vel)
f.beep("bip bip")
```

```python
class Perrari(Auto):
    
    def beep(self, sound="beeeep!"):
        return sound
```

```python
p = Perrari("gialla", 1979)
print(p.beep())
```

# getter e setter / incapsulamento

```python
class Compo():
    def __init__(self):
        self.i = 0
        self.v = 0
        self.temp = 20

c = Compo()
c.v = 12
print("V:", c.v)
```

```python
class Compo2():
    def __init__(self):
        self.i = 0
        self.v = 0
        self.temp = 20

    def getV(self):
        return self.v
    
    def setV(self, v):
        self.v = v

c = Compo2()
c.setV(10)
print(c.getV())
```

```python
class Compo3():
    def __init__(self):
        self.__v = 0
    
    def getV(self):
        return self.__v
    
    def setV(self, v):
        self.__v = v
        
    
c = Compo3()
```

```python
class Compo4():
    def __init__(self):
        self.__v = 0
    
    def getV(self):
        return self.__v
    
    def setV(self, v):
        self.__v = v

    voltage = property(fget=getV, fset=setV)    
    
c = Compo3()
c.voltage = 12
print(c.voltage)
```

decorators

```python
class Compo4():
    def __init__(self):
        self.__v = 0
    
    @property
    def voltage(self):
        return self.__v
    
    @voltage.setter
    def voltage(self, v):
        self.__v = v

c = Compo4()
c.voltage= 20
print(c.voltage)
```
