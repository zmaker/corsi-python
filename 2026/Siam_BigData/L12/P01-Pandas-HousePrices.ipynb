## Amsterdam DataSet

```python
import pandas as pd
```

```python
#apro il file con i dati
df = pd.read_csv("HousingPrices-Amsterdam-August-2021.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
#ipotesi 1 - elimino le righe
df1 = df.dropna(subset=['Price'])
df1.info()
```

```python
#ipotesi2 - inserisco il valor medio
valmedio = df["Price"].mean()
print("valore medio: ", valmedio)
df2 = df.copy()
df2["Price"] = df2["Price"].fillna(valmedio)
df2.info()
```

Proseguo usando il dataframe df1 - quello a cui ho eliminato le righe

```python
#rinomino una colonna
df10 = df1.rename(columns={"Unnamed: 0":"houseId"})
df10.head()
```

```python
import matplotlib as plt
df10["Price"].plot(kind="hist", bins=50, edgecolor="black")
```

```python
df10.plot(kind="scatter", x="Area", y="Price", color='blue', alpha=0.2)
```

```python
df11 = df10[ df10["Price"] < 500000]
df11.plot(kind="scatter", x="Area", y="Price", color='blue', alpha=0.2)
```

```python
#aggiungo una colonna con il prezzo al mq
df10["sqrtprz"] = df10["Price"]/df10["Area"]
df10.head()
```

```python
df10.plot(kind="scatter", x="Area", y="Price", color="blue", alpha=0.1, s=df10["sqrtprz"]/100)
```

```python
df10.plot(kind="scatter", x="Lat", y="Lon", color='blue', alpha=0.2)
```
