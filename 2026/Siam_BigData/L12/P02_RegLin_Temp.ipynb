## Regressione lineare semplice 

Predirre le temperatura. Conversione da C a F

```python
import sklearn
```

```python
import pandas as pd
import numpy as np
```

```python
#genero un array di numeri che rappresentano le temperature in celsius
temp_c = np.random.uniform(low=-10, high=100, size=20)
#riduco i decimali
temp_c = np.round(temp_c, 2)
print(temp_c)
```

```python
#ricavo i gradi farhenheit
temp_f = (temp_c * 1.8) + 32.0
temp_f = np.round(temp_f, 2)
print(temp_f)
```

```python
nn = np.random.random(20)
nn = (nn * 4.0) - 2.0 
print(nn)
```

```python
temp_f = temp_f + nn
temp_f = np.round(temp_f, 2)
print(temp_f)
```

```python
import matplotlib.pyplot as plt
plt.scatter(temp_c, temp_f)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
#i dati son osu una riga ma io ho bisogno di una colonna
print(temp_c.shape)
```

```python
X = temp_c.reshape(-1, 1)
print(X.shape)
print(X)
```

```python
y = temp_f 
```

```python
model.fit(X, y)
```

```python
model.predict([[0]])
```

```python
x_prev = np.array([[110.0],[120.0],[130.0],[140.0],[150.0]])
y_prev = model.predict(x_prev)
```

```python
print(x_prev)
print(y_prev)
```

```python
plt.scatter(X, y)
plt.scatter(x_prev, y_prev)
```
