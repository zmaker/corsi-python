filtri

```python
n1 = np.random.randint(0,100, 20)
print(n1)
nf = n1[n1 < 50]
print(nf)
```

```python
#i numeri pari
pari = n1[n1%2 == 0]
pari
```

```python
#numeri in un intervallo
x = n1[ (n1>30) & (n1<50) ] #numeri di n1 che sono > di 30 E minori di 50
x
```

```python
#numeri in un intervallo
x = n1[ (n1<30) | (n1>50) ] #numeri di n1 che sono < di 30 O > di 50
x
```

```python
x = np.where( (n1%2 == 0), "P", "D" )
x
```

# Files

```python
lista = np.genfromtxt("numeri.txt")
print(lista)
```

```python
lista.dtype
```

```python
lista.size
```

```python
print("min", np.amin(lista))
print("max", np.amax(lista))
print("media", np.average(lista))
print("dev std", np.std(lista))
```

```python
lista = lista * 1.10
lista
```

```python
np.savetxt("numeri_mod.txt", lista, fmt="%0.2f")
```

```python
np.save("lista.dat", lista)
dati = np.load("lista.dat.npy")
dati
```
