# Insiemi - Set
Raccolgono gruppi di dati ma non sono ordinati, non hanno un indice. Non ci possono essere duplicati. Posso però aggiungere elementi o toglierli.

```python
s1 = {'a', 'b', 'c', 'd'}
print(s1)
```

```python
s2 = {'a', 'b', 'c', 'd', 'a', 'b'}
print(s2)
```

```python
v = 'c' in s2
print(v)
```

aggiungo elementi con add()

```python
s2.add('z')
print(s2)
```

rimuovo un elemento con remove() o discard()

```python
s2.discard('z')
print(s2)
```

```python
s2.clear()
print(s2)
```

Operazioni con insiemi 

```python
A = {1,2,3,4,5}
B = {4,5,6,7,8}
C = A.union(B)
print(C)
```

```python
I = A.intersection(B)
print(I)
```

```python
D1 = A.difference(B)
print(D1)
D2 = B.difference(A)
print(D2)
```

```python
DS = A.symmetric_difference(B)
print(DS)
```

# Dizionari 
Contengono dati organizzati in coppie chiave-valore

```python
listino = {'mele':12, 'pere':23, 'kiwi':45}
print(listino)
```

```python
val = listino['mele']
print(val)

#accesso con get
v2 = listino.get('pere')
print(v2)
```

```python
listino['mele'] = 15
print(listino)
```

```python
listino['fragole'] = 99
print(listino)
```

```python
lista_chiavi = listino.keys()
print(lista_chiavi)
```

```python
flag = 'fragole' in listino
print(flag)
```

```python
listino.pop('fragole')
print(listino)
```

```python
for x in listino:
    print(x)
```

```python
print(listino)
listino2 = listino

listino2['patate'] = 66
print(listino2)
print(listino)
```

```python
listino3 = listino.copy()
listino3['carote'] = 55

print(listino3)
print(listino)
```
