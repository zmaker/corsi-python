```python
def multicalc(a, b):
    s = a + b
    p = a * b
    return s, p

ret = multicalc(10,20)
print(ret)

somma, prodotto = multicalc(10,20)
print(somma)
print(prodotto)
```

Scope delle variabili - visibilità

```python
a = 10

def f1():
    a = 20
    print("in f1 a vale:", a)

print(a)
```

```python
f1()
print(a)
```

```python
def f2():
    global a
    a = 20
    print("in f2 a vale:", a)

print(a)
f2()
print(a)
```

```python
def f3(a):
    a = 30
    print("in f3 a vale:", a)

print(a)
f3(a)
print(a)
```

```python
l = [1,2,3]
def f4(ll):
    ll[0] = 99

print(l)
f4(l)
print(l)
```
