## JSON

```python
import json
```

```python
with open("dati.json") as f:
    doc = json.load(f)
    print(doc)
```

```python
with open("dati.json") as f:
    doc = json.load(f)
    print(doc["id"])
    print(doc["type"])
    print(doc["batters"])
    lista = doc["batters"]["batter"]
    print(lista)
    for el in lista:
        print(el["type"])
```

Creazione di un file JSON a partire da oggetti Python

```python
doc = {"nome":"mario", "cod":123, "data":"01/01/2025", "materie":[{"mat":"ita"}, {"mat":"eng"}, {"mat":"gin"}] }
with open("primo.json", "w") as f:
    json.dump(doc, f)
```
