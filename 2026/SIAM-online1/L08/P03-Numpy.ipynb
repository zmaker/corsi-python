# filtri

```python
n = np.random.randint(0, 100, 20)
print(n)

nf = n[ n < 50 ]
print(nf)
```

```python
nf = n[ n == 2 ]
print(nf)
```

```python
# != significa DIVERSO DA
nf = n[ n != 0]
print(nf)
```

```python
# i numeri pari
npa = n[ (n%2) == 0 ]
print(npa)
```

```python
# i numeri compresi in un certo intervallo
# AND -> "e" -> &
ni = n[ (n>20) & (n<60) ]
print(ni)
```

```python
# OR -> "oppure" -> |
ni = n[ (n<20) | (n>60)]
print(ni)
```

# where

```python
print(n)
nn = np.where( n>40, n, 0 )
print(nn)
```

```python
nn = np.where( (n%2 == 0), 'P', 'D' )
print(nn)
print(np.where((n%2 == 0)))
```
