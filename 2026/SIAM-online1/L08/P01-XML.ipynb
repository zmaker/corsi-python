## XML 
Utilizzo il modello DOM (Doc Object Model). Tutto il file è interpretato e trasformato in un tree di oggetti.

```python
from xml.dom import minidom
doc = minidom.parse("biblio.xml")
```

```python
print(doc.firstChild.tagName)
```

```python
books = doc.getElementsByTagName("book")
print(books)
print(len(books))
```

```python
for b in books:
    bid = b.getAttribute("id")
    print(bid)
```

```python
autore = books[0].getElementsByTagName("author")
print(autore)
name = autore[0].firstChild.nodeValue
print(name)
```
