## Statistiche

```python
import statistics as st

voti = [4,5,4,6,7,8,6,7,5,6,9,8,7]
```

```python
#media dei voti
n = st.mean(voti)
print("media: ", n)
```

```python
#massimo e monimo di una lista:
n = min(voti)
print("minimo: ", n)
n = max(voti)
print("massimo: ", n)
```

```python
#mediana
n = st.median(voti)
print("mediana: ", n)

voti.sort()
print(voti, len(voti))
```

```python
n = st.variance(voti)
print(n)
```

```python
n = st.stdev(voti)
print(n)
```
