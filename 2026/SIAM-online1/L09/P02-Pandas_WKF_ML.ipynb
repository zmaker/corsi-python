## Workflow Ripulitura Mailing List CSV

```python
import pandas as pd
```

```python
df = pd.read_csv("dati_mailing_list.csv", delimiter=",")
df.head()
```

Esploro i dati nel file usando info, describe, dtype, hist...

```python
df.info()
```

```python
df.describe()
```

```python
df.dtypes
```

Modifico i nomi delle colonne

```python
df = df.rename(columns={"__Email_Entered":"email"})
df.head()
```

```python
df.rename(columns={"First name":"nome"}, inplace=True)
df.head()
```

```python
df.rename(columns={"Surname":"cognome", "Gender":"sesso", "Age":"eta", "_CREDIT":"credito"}, inplace=True)
df.head()
```

Devo eliminare le colonne Logins e IP_Address. Posso dare un drop (axis=1), oppure seleziono le colonne che voglio tenere. In pratica creo una sotto-tabella

```python
df2 = df[["email", "nome", "cognome", "sesso", "eta", "credit"]].copy()
df2.head()
```

Ora rimuovo gli spazi in eccesso (strip()) nelle colonne "nome" e "cognome"

```python
df2["nome"] = df2["nome"].str.strip()
df2["nome"] = df2["nome"].str.title()
df2.head()
```

```python
df2["cognome"] = df2["cognome"].str.strip().str.title()
df2.head()
```

Elimino le email non valide usando una RegExp. 

```python
import re
email_re = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$"
```

```python
df3 = df2[ df2["email"].str.match(email_re, na=False) ].copy()
df3
```

Elimino le righe duplicate

```python
df3.duplicated().sum()
```

```python
df4 = df3.drop_duplicates(subset=["email"], keep="first").copy()
df4.head()
```

Tratto le colonne con i dati mancanti. 

```python
#1. calcolo l'eta media
etamedia = df4["eta"].mean()
print(etamedia)

#2. inserisco il valore solo dove manca
df4["eta"] = df4["eta"].fillna(etamedia)
df4.head()
```

```python
df4["eta"] = df4["eta"].round(0).astype(int)
df4
```

```python
df4.dtypes
```

Seleziono le righe con credito > 0

```python
df5 = df4[ df4["credit"] > 0 ]
df5
```

Salvo il file

```python
df5.to_csv("ml_clean.csv", index=False)
```
