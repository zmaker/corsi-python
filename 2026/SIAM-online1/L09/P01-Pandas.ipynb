## indici e posizioni
due concetti importanti:
* chiave della riga
* posizione della riga

**loc[]** è una funzione che estrae una riga usando l'**indice** (chiave primaria data da Pandas)

```python
r = tab2.loc[0]
r
```

```python
# iloc estrae una riga usando la sua posizione
r = tab2.iloc[0]
r
```

```python
print("nome: ", r["nome"])
```

Genero una dataframe specificando l'indice, i dati e i nomi delle colonne

```python
indici = ["01", "02", "03", "04"]
lista = ["mario", "luigi", "anna", "stefano"]
df = pd.DataFrame(data=lista, index=indici, columns=["nomi"])
df
```

```python
r = df.loc["02"]
print(r)
```

Creo un DataFrame specificando più colonne

```python
indici = [10,20,30,40]
lista = [
    ["mela", 100, 1.98],
    ["pera", 56, 2.34],
    ["kiwi", 45, 6.34],
    ["fragola", 78, None]
]
colonne = ["prodotto", "qta", "prz"]
df = pd.DataFrame(data=lista, index=indici, columns=colonne)
df
```

Genero un indice con un range

```python
lista = ["mario", "luigi", "anna", "stefano"]
df = pd.DataFrame(data=lista, index=range(1,5))
df
```

## Filtraggio dati

```python
tab
```

```python
#applico un filtro e visualizzo solo le righe con sesso = 'M'
df1 = tab[ tab.sesso == 'M' ]
df1
```

```python
tab[ tab.sesso == 'F']
```

```python
#seleziono le persone con altezza superiore a 1.80
tab[ tab.altezza > 1.80]
```

Posso combinare più espressioni logiche usando gli operatori:
* AND -> &
* OR -> |
* NOT -> ~

```python
tab[ (tab.altezza > 1.80) & (tab.sesso == 'F') ]
```

```python
tab[ (tab.altezza < 1.70) | (tab.altezza > 1.85) ]
```

```python
tab[ ~(tab.nome == 'Tina') ]
```

```python
tab[ ~(tab.nome == 'Tina') & (tab.altezza > 1.70) ]
```
