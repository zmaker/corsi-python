```python
from openpyxl import load_workbook
```

```python
#apro il file excel
wb = load_workbook("caseroma.xlsx")

#stampo i fogli di lavoro 
print(wb.sheetnames)
```

Seleziono un foglio su cui lavorare

```python
# ws = wb['Historical Rents']
#lista delle pagine (worksheet)
pagine = wb.sheetnames

#seleziono la prima pagina del foglio
ws = wb[pagine[0]]
```

Leggo il valore di una cella

```python
valore = ws["B2"].value
print(valore)
```

Leggo una riga

```python
riga = [] #creo una lista vuota per contenere i valori della riga selezionata
for cella in ws[2]: #scorre le celle dalla riga 2
    riga.append(cella.value)
    
print(riga)
```

Modifico il valore di una cella

```python
ws["B2"] = "Vaticano"
```

Scrivo o salvo il file

```python
wb.save("out.xlsx")
```
