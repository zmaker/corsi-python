```python
import pandas as pd
```

```python
# bins = 30 -> crea 30 suddivisioni

df.hist(edgecolor='black', bins=30, figsize=(12, 8))
```

Nel file ci sono delle righe con prezzo non valorizzato (NaN). Cosa faree? Come li tratto?

```python
#caso 1 - elimino le righe non valorizzate
df1 = df.dropna( subset=['Price'] )
df1.info()
```

```python
#Caso 2 - inserisco il valor medio del Prezzo
valmedio = df["Price"].mean()
print("valmedio: ", valmedio)

#creo una copia effettiva del df
df2 = df.copy()
df2["Price"] = df2["Price"].fillna(valmedio)
df2.info()
```

```python
#rinomino la prima colonna che non ha un nome vero e proprio
df3 = df1.rename( columns={ "Unnamed: 0":"houseId" } )
df3.head()
```

Provo a creare un istogramma dei prezzi

```python
import matplotlib as plt

df3["Price"].plot(kind="hist", bins=50, edgecolor="black")
```

```python
df3.plot(kind="scatter", x="Area", y="Price", alpha=0.1)
```

Se volessi vedere una particolare zona dei dati, come posso fare? DEvo aplicare dei filtri

```python
df4 = df3[ df3["Price"] < 500000 ]
df4.plot(kind="scatter", x="Area", y="Price", alpha=0.1)
```

```python
#aggiungo una colonna con il prezzo al mq e poi la userò per modificare la dimensione dei pallini
df3["sqrtprz"] = df3["Price"]/df3["Area"]
df3.head()
```

```python
df3.plot(kind="scatter", x="Area", y="Price", alpha=0.05, s=df3["sqrtprz"]/100)
```

```python
df3.plot(kind="scatter", x="Area", y="Price", alpha=0.05, s=df3["Room"]*10)
```

```python
df3.plot(kind="scatter", x="Lat", y="Lon", alpha=0.1)
```

Proviamo a mettere i dati su una maps di Google
https://python-visualization.github.io/folium/latest/getting_started.html



```python
#%pip install folium
import folium
```

```python
mappa = folium.Map(location=[52.36, 4.85], zoom_start=11, width='70%', height='60%')
for index, row in df3.iterrows():
    folium.CircleMarker(
        location=[row['Lat'], row['Lon']],
        radius = row["Price"]/100000,
        color='blue',
        fill=True,
        fill_color='blue',
        fill_opacity = 0.2,
        tooltip=row['Price']
    ).add_to(mappa)
mappa
```
