**forma compatta per usare i file**

```python
with open("file1.txt") as f:
    print(f.read())
```

**Come scrivere file di testo**
devo aggiungere a open() il parametro "w" che significa write = scrivi

\n per andare a capo

```python
with open("file2.txt", "w") as f:
    f.write("pane\n")
    f.write("latte\n")
    f.write("burro\n")
```

per scrivere con append uso il flag "a". Con append aggiungo sempre in coda al file

```python
with open("file3.txt", "a") as f:
    f.write("ciao\n")
```
