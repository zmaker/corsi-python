### Insiemi / Set

```python
s1 = {2, 4, 6, 8, 10}
print(s1, type(s1))
```

```python
l1 = [1,2,2,2,2,3,3,4,5,6,7,8,8,8,8,10]
print(l1)
s2 = set(l1)
print(s2)
```

Gli insiemi sono modificabili

```python
s3 = set()
s3.add("mela")
s3.add("pera")
s3.add("fragola")
s3.add("cipolla")
print("s3", s3)
```

```python
s3.discard("cipolla")
print(s3)
```

```python
if "banana" in s3:
    print("banana ok")
else:
    print("banana ko")
```

```python
for el in s3:
    print(el)
```

### Operazioni su insiemi

```python
a = {1,2,3,10,20}
b = {10,20,30,40,50}
```

```python
c = a.union(b)
print(c)
```

```python
d = a.intersection(b)
print(d)
```

```python
e = a.difference(b)
print(e)
```

```python
f = b.difference(a)
print(f)
```

```python
g = a.symmetric_difference(b)
print(g)
```
