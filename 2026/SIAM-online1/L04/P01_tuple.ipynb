### Tuple
Sono come liste ma a sola lettura

```python
frutta = ("mele", "pere", "kiwi")
print(frutta)
```

Posso accedere agli elementi tramite un indice

```python
print(frutta[0])
```

Per trovare il numero di elementi uso len()

```python
print(len(frutta))
```

Scorrere una tupla con un ciclo for

```python
for el in frutta:
    print(el)
```

**Packing/unpacking**

```python
dati = ("mela", 100, 1.0)
a,b,c = dati
print("prodotto:", a)
print("n.pezzi:", b)
print("prezzo: ", c)
```
