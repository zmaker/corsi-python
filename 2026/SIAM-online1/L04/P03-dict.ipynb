### Dizionari / Dict

```python
listino = { "mele":12, "pere":23, "kiwi":45 }
print(listino)
```

```python
n = listino["mele"]
print(n)
```

```python
listino["mele"] = 99
print(listino)
```

```python
listino["fragole"] = 100
print(listino)
```

```python
if "fragole" in listino:
    print("fragole ok")
```

```python
for el in listino.keys():
    print(el)
```

```python
for el in listino.values():
    print(el)
```

```python
for el in listino.items():
    #print(el)
    prod, qta = el
    print(prod, qta)
```

**copia di dict**

```python
listino2 = listino
print("l", listino)
print("l2", listino2)
```

```python
listino2["mele"] = 0
print("L", listino)
print("L2", listino2)
```

```python
listino3 = listino.copy()
print("L", listino)
print("L3", listino3)
```

```python
listino3["mele"] = 100
print("L", listino)
print("L3", listino3)
```
