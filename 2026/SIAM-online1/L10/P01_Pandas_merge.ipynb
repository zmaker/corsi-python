## Pandas - Join/merge

Le join si effettuano con il metodo merge 

```python
import pandas as pd
```

```python
#creo un dict con due voci:
data = {'codice':range(1001, 1006), 'nome':['mela', 'pera', 'banana', 'kiwi', 'fragola']}
df_prodotti = pd.DataFrame(data)
df_prodotti
```

```python
#creo una seconda tabella con le qta
data = {'codice':range(1001, 1011), 'qta':[10,20,30,40,50,60,70,80,90,99]}
df_magazzino = pd.DataFrame(data)
df_magazzino
```

```python
df1 = pd.merge(df_prodotti, df_magazzino, on='codice', how='left')
df1
```

```python
df2 = pd.merge(df_prodotti, df_magazzino, on='codice', how='right')
df2
```

```python
#aggiungo una riga a df_prodotti
df_prodotti.loc[5] = {"codice":1100, "nome":"fichi"}
df_prodotti
```

```python
df2 = pd.merge(df_prodotti, df_magazzino, on='codice', how='right')
df2
```

```python
df2 = pd.merge(df_prodotti, df_magazzino, on='codice', how='inner')
df2
```

```python
df2 = pd.merge(df_prodotti, df_magazzino, on='codice', how='outer')
df2
```

## Concat
Serve per unire due dataframe con struttura compatibile (ugual numero e nome di colonne)

```python
data = {'codice':[10,20,30],
        'nome':['kiwi','mela','pera'],
        'qta':[100, 67, 34]}
dfa = pd.DataFrame(data)
dfa
```

```python
data = {'codice':[100,120],
        'nome':['patate','carote'],
        'qta':[22,33]}
dfb = pd.DataFrame(data)
dfb
```

```python
dfc = pd.concat([dfa, dfb], ignore_index=True)
dfc
```

caso in cui unisco 2 tabelle e rigenero l'indice in seguito

```python
dfd = pd.concat([dfa, dfb])
dfd
```

```python
#rigenero l'indice
dfd.reset_index(drop=True, inplace=True)
dfd
```

## Raggruppamenti - GroupBy

```python
data = {
    'codice':range(1001, 1006), 
    'nome':['mela', 'pera', 'banana', 'kiwi', 'fragola'],
    'qta':[12,23,34,45,56],
    'id_mag':[1,0,1,1,0]
}
dfp = pd.DataFrame(data)
dfp
```

```python
dfp.groupby("id_mag").groups
```

```python
dfp.groupby("id_mag").get_group(0)
```

```python
dfp.groupby("id_mag").get_group(1)
```

```python
dfp.groupby("id_mag")['qta'].sum()
```

```python
dfp.groupby("id_mag")["qta"].describe()
```
