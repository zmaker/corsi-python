```python
# %pip install matplotlib

import matplotlib.pyplot as plt
```

```python
y = [2,3,4,3,4,6,3,4,3,4,5,6,7]
plt.plot(y)
```

```python
plt.plot([10,20,25,50],[1,4, 10, 16], 'ro')
plt.show()
```

```python
#rappresento più serie di dati sullo stesso grafico
import numpy as np

t = np.arange(0., 5., 0.2)

plt.plot(t,t, "r--", t, t**2, 'bs', t, t**3, 'go')
```

```python
plt.figure() #creo un'immagine in cui inserirò più grafici
plt.subplot(211) #delimito l'area in cui mettere il primo grafico
# 211 = 2 righe - 1 colonna - mi metto nella prima pos
plt.plot([1,2,3,4],[1,4,9,16], 'b-')

plt.subplot(212) #delimito l'area in cui mettere il secondo grafico
# 212 = 2 righe - 1 colonna - mi metto nella seconda pos
plt.plot([1,2,3,4],[4,6,7,9], 'r-')

plt.show()
```

```python
#codice copiato dal sito matplotlib

data = {'apple': 10, 'orange': 15, 'lemon': 5, 'lime': 20}
names = list(data.keys())
values = list(data.values())

fig, axs = plt.subplots(1, 3, figsize=(9, 3), sharey=True)
axs[0].bar(names, values)
axs[1].scatter(names, values)
axs[2].plot(names, values)
fig.suptitle('Categorical Plotting')
```
