```python
import numpy as np

# genero una lista di mesi
mesi = ["ge", 'fe', 'ma', 'ap', 'mg', 'gi', 'lu', 'ag', 'st', 'ot', 'nv', 'di']

#genero dei valori casuali per simulare i movimenti in ingresso/uscita di un conto corrente
ingressi = np.random.randint(1000, 5000, size=12)
uscite = np.random.randint(1000, 5000, size=12)

#creo il dataframe
cc = pd.DataFrame( {'mese':mesi, "entrate":ingressi, "uscite":uscite} )
cc
```

```python
# %pip install xlsxwriter
```

```python
wr = pd.ExcelWriter("conto.xlsx", engine='xlsxwriter')
cc.to_excel(wr, index=False, sheet_name="conto")
#operazioni sul file excel
wr.close()
```

```python
with pd.ExcelWriter("conto2.xlsx", engine='xlsxwriter') as wr:
    cc.to_excel(wr, index=False, sheet_name="conto")

    # wr.book mi permette di lavorare sul modello del file excel
    workbook = wr.book
    #seleziono il tab del mio foglio excel
    ws = wr.sheets['conto']
    #modifico una cella
    f1 = workbook.add_format({'bold':True, 'font_color':'blue'})
    ws.write("D2", "Hello!", f1)
```

## Pandas e Database

Consigliato accesso con https://www.sqlalchemy.org/

```python
# %pip install sqlalchemy
```

```python
from sqlalchemy import create_engine
#creo la connessione al database e nel caso lo crea
eng = create_engine('sqlite:///banca.db')
# trasferisce il dataframe in una tabella
cc.to_sql('conto', eng, index=False)
```

```python
#lettura dati da una tabella
with eng.connect() as c:
    df2 = pd.read_sql_table("conto", c)

df2.head()
```

```python
with eng.connect() as c:
    df3 = pd.read_sql_query("SELECT * FROM conto WHERE entrate > 2500", c)

df3
```

```python
#chiude la conenssione al file
eng.dispose()
```
