## Lotteria Pandas

```python
import numpy as np

numeri = np.arange(1, 91)
numeri
```

```python
#mescolo le palline
np.random.shuffle(numeri)
numeri
```

```python
#creo una lista numpy vuota in cui inseriro le estrazioni

estrazioni = np.array([])
```

```python
for i in range(20):
    #rimescolo i numeri
    np.random.shuffle(numeri)

    #estraggo 5 numeri
    estratti = numeri[0:5]
    #print(estratti)

    #li aggiungo alla lista di estrazioni
    estrazioni = np.append(estrazioni, [estratti])

estrazioni = estrazioni.reshape(20, 5)
estrazioni
```

```python
import pandas as pd

#creo un intervallo di date con pandas
dates = pd.date_range('20260101', periods=20)
dates
```

```python
df = pd.DataFrame(data=estrazioni, index=dates)
df
```

```python
for i in range(len(df)):
    print(df.iloc[i])
```

```python
ritardatari = {}
for i in range(len(df)):
    riga = df.iloc[i]
    for num in riga:
        gg = len(df)-i
        #print(num, gg)
        ritardatari[num] = gg

ritardatari
```

```python
df = pd.DataFrame(list(ritardatari.items()), columns=['Valore', 'Conteggio'])
df.head()
```

```python
df = df.sort_values(by="Conteggio", ascending=False)
df.head(10)
```
