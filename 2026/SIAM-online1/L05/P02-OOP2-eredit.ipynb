### Ereditarietà

```python
class Auto():
    def __init__(self, colore, anno):
        self.colore = colore
        self.anno = anno
        self.speed = 0

    def setSpeed(self, spd):
        self.speed = spd
    
    def beep(self, suono):
        return f"{suono})))"
    
    def __str__(self):
        return f"Auto: c:{self.colore} a:{self.anno} v:{self.speed} "
```

```python
class Fiat500(Auto):
    pass
```

```python
a1 = Auto("blu", 1997)
a1.setSpeed(100)
print(a1)
```

```python
a2 = Fiat500("giallo", 1967)
print(a2.beep("biii"))
print(a2)
```

```python
class Ferrari(Auto):
    def __init__(self, colore, anno, turbo):
        super().__init__(colore, anno)
        self.turbo = turbo
    
    def beep(self, suono):
        return f"({suono}!!!!!!)"
    
    def __str__(self):
        return f"Ferrari: c:{self.colore} a:{self.anno} v:{self.speed} t:{self.turbo}"
    

f1 = Ferrari("rosso", 2001, True)
print(f1)
print(f1.beep("UUUUU"))
```
