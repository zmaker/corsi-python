### costruttore
una funzione speciale chiamata quando istanzio la classe

```python
class Pera():
    def __init__(self):
        #costruttore della classe Pera
        print("Pera creata!")
```

```python
p1 = Pera()
```

aggiungiamo delle proprietà

```python
class Banana():
    def __init__(self):
        self.peso = 1000
        self.colore = "giallo"

    def getPeso(self):
        return self.peso
    
    def setPeso(self, p):
        if p < 0:
            self.peso = 0
        else:
            self.peso = p
```

```python
b1 = Banana()
b1.peso = 100

print(b1.peso)

p = b1.getPeso()
print(p)

b1.setPeso(120)
print(b1.peso)
```

```python
class Patata():
    def __init__(self, peso=1000, colore="marrone"):
        self.peso = peso
        self.colore = colore

p0 = Patata(175, "beige")
p1 = Patata(1250)
p2 = Patata()

print("peso: ", p0.peso, "c: ", p0.colore)
print("peso: ", p1.peso, "c: ", p1.colore)
print("peso: ", p2.peso, "c: ", p2.colore)
```

```python
class Carota():
    def __init__(self, peso, lun, colore="arancione"):
        self.peso = peso
        self.lunghezza = lun
        self.colore = colore
    
    def __str__(self):
        return f"Carota {self.peso}g {self.lunghezza}cm di color {self.colore}"

c1 = Carota(123, 15)
c2 = Carota(lun=18, peso=77.7)
c3 = Carota(235, 17, "verde")

print(c1)
```

```python
class Punto():
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def __str__(self):
        return f"({self.x};{self.y})"
    
    def __del__(self):
        #distruttore - chiamato quando l'oggetto è scaricato dalla memoria
        print("bye bye")
    
p1 = Punto(10,23)
print(p1)

#scarico l'oggetto in modo esplicito
del p1
```

```python
class Zucchina():
    def __init__(self, lun, peso, col="verde"):
        self.peso = peso
        self.lunghezza = lun
        self.colore = col
        self.sapore = "erbaceo"
        self.fette = 0
    
    def __str__(self):
        txt = f"Z: {self.lunghezza}cm, {self.peso}g, C:{self.colore} S:{self.sapore}"
        return txt

    def sbuccia(self):
        self.peso = self.peso * 0.9
    
    def affetta(self, ft):
        self.fette = ft
    
z1 = Zucchina(14.5, 110)
print(z1)
z1.sbuccia()
print(z1)
z1.affetta(10)
print(z1.fette)

```

## esercizio biblioteca

```python
class Libro():
    def __init__(self, titolo, autore):
        self.titolo = titolo
        self.autore = autore
    def __str__(self):
        return f"L: {self.titolo} di {self.autore}"
```

```python
class Biblioteca():
    def __init__(self):
        self.scaffale = []
    
    def addBook(self, titolo, autore):
        l = Libro(titolo, autore)
        self.scaffale.append(l)
    
    def __str__(self):
        txt = ""
        for b in self.scaffale:
            txt += str(b) + "\n"
        return txt

    def search(self, titolo):
        ans = Libro("???", "???")
        for b in self.scaffale:
            if b.titolo == titolo:
                ans = b
                break
        return ans
```

```python
b = Biblioteca()
b.addBook("Promessi Sposi", "Manzoni")
b.addBook("Inferno", "Dante")
b.addBook("Python per tutti", "Monty Python")

print(b)

book = b.search("Inferno")
print(book)
book = b.search("Milano Segreta")
print(book)
```
