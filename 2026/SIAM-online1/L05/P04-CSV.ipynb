```python
import csv
```

```python
with open("dati.csv") as f:
    rows = csv.reader(f, delimiter=",")
    r1 = next(rows) #estrae la riga e passa a puntare la successiva
    print(r1)
    print() #stampo una riga vuota
    for rr in rows:
        #print(rr)
        codice = rr[0]
        prodotto = rr[1]
        quantita = rr[2]
        prezzo = rr[3]
        print(f"[{codice}]: {prodotto} n.{quantita} {prezzo} €")
```

Calcolo il numero di prodotti totali nel magazzino

```python
qta_totale = 0
val_totale = 0

with open("dati.csv") as f:
    rows = csv.reader(f, delimiter=",")
    r1 = next(rows) #estrae la riga e passa a puntare la successiva
    print(r1)
    print() #stampo una riga vuota
    for rr in rows:
        #print(rr)
        codice = rr[0]
        prodotto = rr[1]

        quantita = int(rr[2])
        qta_totale += quantita

        prezzo = float(rr[3])
        val_totale += quantita * prezzo
        print(f"[{codice}]: {prodotto} n.{quantita} {prezzo} €")

print("qta totale: ", qta_totale)
print("val totale: ", val_totale)
print(f"val medio: {(val_totale/qta_totale):.2f} eur")
```

**Scrivere file CSV**

```python
with open("magazzino.csv", "w") as f:
    cur = csv.writer(f, delimiter=',', lineterminator='\n')
    cur.writerow(['cod', 'nome', 'cat', 'prezzo'])
    cur.writerow([1, 'stratocaster', 'guitar', 1234.0])
    cur.writerow([2, 'yamaha cs1', 'keyboard', 2345.5])
    cur.writerow([3, 'fender', 'bass', 678.8])
```
